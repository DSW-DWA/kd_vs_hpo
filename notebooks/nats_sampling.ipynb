```python
import xautodl
from xautodl.models import get_cell_based_tiny_net
import json
from nats_bench import create
import numpy as np
import pandas as pd
```

```python
api = create("data/NATS-tss-v1_0-3ffb9.pickle.pbz2", "tss", fast_mode=False, verbose=False)
print("Total architectures:", len(api))
```

```python
rng = np.random.default_rng(42)
model_indexes = rng.integers(0, len(api), size=10)
dataset = 'cifar10'
configs = []

for model_index in model_indexes:
    config = api.get_net_config(model_index, dataset)
    config['num_classes'] = 10
    config["arch_index"] = int(model_index)
    config["search_space"] = "tss"
    config["dataset"] = dataset 
    configs.append(config)
with open ("experiments/nats_architectures_10_additional.json", "w+") as f:
    json.dump(configs, f, indent=4)
```
